```python
import numpy as np
import skvideo.io
from numpy.linalg import svd
from skvideo.datasets import bigbuckbunny
from All import fold, unfold, Mul_3, Mul_2, Mul_1
```

```python
videodata = skvideo.io.vread("original.mp4")
R = videodata[:, :, :, 0]
G = videodata[:, :, :, 1]
B = videodata[:, :, :, 2]
```

```python
videodata.shape
```

```python
print(f"R shape is {R.shape}\nG shape is {G.shape}\nB shape is {B.shape}\n")
```

# Unfolding of Red Channel

```python
unfold_1_R = unfold(R, 1)
unfold_2_R = unfold(R, 2)
unfold_3_R = unfold(R, 3)
```

```python
print(f"unfold_1_R shape is {unfold_1_R.shape}\nunfold_2_R shape is {unfold_2_R.shape}\nunfold_3_R shape is {unfold_3_R.shape}\n")
```

# Unfolding of Green Channel

```python
unfold_1_G = unfold(G, 1)
unfold_2_G = unfold(G, 2)
unfold_3_G = unfold(G, 3)
```

```python
print(f"unfold_1_G shape is {unfold_1_G.shape}\nunfold_2_G shape is {unfold_2_G.shape}\nunfold_3_G shape is {unfold_3_G.shape}\n")
```

# Unfolding of Blue Channel

```python
unfold_1_B = unfold(B, 1)
unfold_2_B = unfold(B, 2)
unfold_3_B = unfold(B, 3)
```

```python
print(f"unfold_1_B shape is {unfold_1_B.shape}\nunfold_2_B shape is {unfold_2_B.shape}\nunfold_3_B shape is {unfold_3_B.shape}\n")
```

# Svd Decomposition of Red Channel

```python
U_1_R, _, _ = svd(unfold_1_R, full_matrices=False)
U_2_R, _, _ = svd(unfold_2_R, full_matrices=False)
U_3_R, _, _ = svd(unfold_3_R, full_matrices=False)
```

```python
print(f"U_1_R shape is {U_1_R.shape}\nU_2_R shape is {U_2_R.shape}\nU_3_R shape is {U_3_R.shape}\n")
```

# Svd Decomposition of Green Channel

```python
U_1_G, _, _ = svd(unfold_1_G, full_matrices=False)
U_2_G, _, _ = svd(unfold_2_G, full_matrices=False)
U_3_G, _, _ = svd(unfold_3_G, full_matrices=False)
```

```python
print(f"U_1_G shape is {U_1_G.shape}\nU_2_G shape is {U_2_G.shape}\nU_3_G shape is {U_3_G.shape}\n")
```

# Svd Decomposition of Blue Channel

```python
U_1_B, _, _ = svd(unfold_1_B, full_matrices=False)
U_2_B, _, _ = svd(unfold_2_B, full_matrices=False)
U_3_B, _, _ = svd(unfold_3_B, full_matrices=False)
```

```python
print(f"U_1_B shape is {U_1_B.shape}\nU_2_B shape is {U_2_B.shape}\nU_3_B shape is {U_3_B.shape}\n")
```

# Compute S Red from U1_R U2_R U3_R

```python
Temp_1_R = Mul_1(U_1_R.T, R)
Temp_2_R = Mul_2(U_2_R.T, Temp_1_R)
S_R = Mul_3(U_3_R.T, Temp_2_R)
```

```python
S_R.shape
```

# Compute S Green from U1_G U2_G U3_G

```python
Temp_1_G = Mul_1(U_1_G.T, G)
Temp_2_G = Mul_2(U_2_G.T, Temp_1_G)
S_G = Mul_3(U_3_G.T, Temp_2_G)
```

```python
S_G.shape
```

# Compute S Blue from U1_B U2_B U3_B

```python
Temp_1_B = Mul_1(U_1_B.T, B)
Temp_2_B = Mul_2(U_2_B.T, Temp_1_B)
S_B = Mul_3(U_3_B.T, Temp_2_B)
```

```python
S_B.shape
```

# Approximate Red Channel Using 20, 100, 200

```python
Apr_U_1_R = U_1_R[:, :20]
Apr_U_2_R = U_2_R[:, :100]
Apr_U_3_R = U_3_R[:, :200]
Apr_S_R = S_R[:20, :100, :200]
```

```python
Apr_temp_1_R = Mul_1(Apr_U_1_R, Apr_S_R)
Apr_temp_2_R = Mul_2(Apr_U_2_R, Apr_temp_1_R)
Reconstructed_Red_Channel = Mul_3(Apr_U_3_R, Apr_temp_2_R)
```

```python
Apr_U_1_G = U_1_G[:, :20]
Apr_U_2_G = U_2_G[:, :100]
Apr_U_3_G = U_3_G[:, :200]
Apr_S_G = S_G[:20, :100, :200]
```

```python
Apr_temp_1_G = Mul_1(Apr_U_1_G, Apr_S_G)
Apr_temp_2_G = Mul_2(Apr_U_2_G, Apr_temp_1_G)
Reconstructed_Green_Channel = Mul_3(Apr_U_3_G, Apr_temp_2_G)
```

# Approximate Blue Channel Using 20, 100, 200

```python
Apr_U_1_B = U_1_B[:, :20]
Apr_U_2_B = U_2_B[:, :100]
Apr_U_3_B = U_3_B[:, :200]
Apr_S_B = S_B[:20, :100, :200]
```

```python
Apr_temp_1_B = Mul_1(Apr_U_1_B, Apr_S_B)
Apr_temp_2_B = Mul_2(Apr_U_2_B, Apr_temp_1_B)
Reconstructed_Blue_Channel = Mul_3(Apr_U_3_B, Apr_temp_2_B)
```

# Concatenate Red Green and Blue channel for 20, 100, 200

```python
Final = np.empty((132, 720, 1280, 3))
Final[:, :, :, 0] = Reconstructed_Red_Channel
Final[:, :, :, 1] = Reconstructed_Green_Channel
Final[:, :, :, 2] = Reconstructed_Blue_Channel

skvideo.io.vwrite("outputvideo_20_100_200.mp4", Final)
```
